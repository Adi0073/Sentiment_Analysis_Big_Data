```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification

MODEL_PATH = r"D:\ecom-sentiment-stream\models\pretrained_models\pretrained_models\distilbert-amazon-reviews-sentiment"

tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_PATH,
    local_files_only=True
)

print("Model loaded successfully!")
print(model.config.id2label)
```

```python
import torch

model.eval()

review = "The product quality is excellent. Delivery was fast and I am very happy."

inputs = tokenizer(
    review,
    return_tensors="pt",
    truncation=True,
    max_length=256
)

with torch.no_grad():
    outputs = model(**inputs)

probabilities = torch.softmax(outputs.logits, dim=1)

predicted_class = torch.argmax(probabilities, dim=1).item()

print("Review:", review)
print("Predicted sentiment:", model.config.id2label[predicted_class])
print("Confidence:", round(probabilities[0][predicted_class].item(), 4))
```

```python
reviews = [
    "Excellent product. Worth every penny.",
    "The product is average. Nothing special.",
    "Terrible quality. Completely disappointed.",
    "Fast delivery and excellent packaging.",
    "The product stopped working after two days."
]

for review in reviews:
    inputs = tokenizer(
        review,
        return_tensors="pt",
        truncation=True,
        max_length=256
    )

    with torch.no_grad():
        outputs = model(**inputs)

    prediction = outputs.logits.argmax(dim=1).item()

    print(review)
    print("Sentiment:", model.config.id2label[prediction])
    print("-" * 60)
```
